# Working with CSV

Let's import a file with fuel economy data

```
import csv

with open('mpg.csv') as csvfile:
    mpg = list(csv.DictReader(csvfile))

print(mpg)
```

Find out how many items are there in the dictionary

```
for item in mpg[1].items():
    print(item)
```

Find the column names with keys

```
print(mpg[0].keys())
```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```
total = 0

for item in mpg:
    total += float(item['cty'])

average = total / len(mpg)

print(average)
```

Find the average hwy fuel economy across all cars

```
total = 0

for item in mpg:
    total += float(item['hwy'])

average = total / len(mpg)

print(average)
```

Use set to return the unique values for the number of cylinders the cars.

```
cylinders = set()

for item in mpg:
    cylinders.add(item['cyl'])

print(cylinders)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```
groups = {}

for item in mpg:
    cyl = item['cyl']
    cty = float(item['cty'])

    if cyl not in groups:
        groups[cyl] = []

    groups[cyl].append(cty)

for cyl in groups:
    average = __builtins__.sum(groups[cyl]) / len(groups[cyl])
    print(cyl, average)
```

Return the unique values for the class types in our dataset using set.

```
class_types = set()

for item in mpg:
    class_types.add(item['class'])

print(class_types)
```

Find average miles per gallon in the highway for each class of vehicles

```
groups = {}

for item in mpg:
    vehicle_class = item['class']
    hwy = float(item['hwy'])

    if vehicle_class not in groups:
        groups[vehicle_class] = []

    groups[vehicle_class].append(hwy)

for vehicle_class in groups:
    average = __builtins__.sum(groups[vehicle_class]) / len(groups[vehicle_class])
    print(vehicle_class, average)
```
